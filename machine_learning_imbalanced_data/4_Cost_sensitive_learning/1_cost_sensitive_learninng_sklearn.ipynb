```python
from sklearn.linear_model import LogisticRegression
```

```python
import sklearn.linear_model
dir(sklearn.linear_model)
```

```python
from sklearn.metrics import roc_auc_score
```

```python
def run_Logit(X_train, X_test, y_train, y_test, class_weight):

    logit = LogisticRegression(
        penalty='l2',
        solver='newton-cg',
        random_state=0,
        max_iter=10,
        n_jobs=-1,
        class_weight=class_weight
    )

    logit.fit(X_train, y_train)

    print('Train set')
    pred = logit.predict_proba(X_train)[:,1]
    print(
        'Logit roc-auc: {}'.format(roc_auc_score(y_train, pred))
    )

    print('Test set')
    pred = logit.predict_proba(X_test)[:,1]
    print(
        'Logit roc-auc: {}'.format(roc_auc_score(y_test, pred))
    )
```
