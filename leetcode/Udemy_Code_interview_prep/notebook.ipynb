```python
class LinkedList:
    def __init__(self, value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.length = 1

    def print_list(self):
        temp = self.head
        while temp is not None:
            print(temp.value)
            temp = temp.next

    def append(self, value):
        
        new_node = Node(value)
        #In case thte linkedlist is an empty list
        if self.length == 0:
            #By assignign the head and tail reference to the newly generate node, we are essnetially apending here for a case of an empty list
            self.head = new_node
            self.tail = new_node
        else:
            #handling in cases where the list is not empty
            #Append the new node to the last node. 
            #This would be the node that the tail points to 
            #Specifically we want to modify so that the next attrtibute of the last no de to point to the new node.
            #Once that is completed then we want the self.tail to point to the new node

            #Appending should take advantage of the tail rfeerence, Assume that we are appending from the tail end that is because as we  know that when we are doing stack, we are appending fromt he head refernce. But in this case we are appending fromt he tail. Hence the appending.
            #That's why in the cases of stac we don't refer to it as appending, innst3ead we are referring to it as pushing, because for one we are trying to emphasize the Last Iin First Out nature.
            #Furthermore, in the case of stakcs they are not horizontal, they appear veritical
            self.tail.next = new_node

            #At this point the tail still points to the last node
            #Have the tail points to the new node
            self.tail = new_node
        
        #Update the length of the LinkedList by 1. Since we are adding a new node
        self.length += 1

    def pop(self):
        #case where the LL is already empty
        #In this case nothing to pop and we return None
        if self.length == 0:
            return None
        
        #case where the LL is not empty 
        #Using temp to label the node that will be popped off
        #Using the pre as a marker for 2nd before last node
        temp = self.head
        pre = self.head 

        #enter while loop to locate the 2nd before last node
        while temp.next is not None:
            
            #The order here matters
            #We want to set the pre to point to the same node the temp points to, if and only if the node that temp points to is not the last node.
            #Given we could get to this point only if temp.next is not None, then we can help pre to point to this node
            pre = temp

            #update the temp to point to the next node
            #the while loop will verify if the next Node is the last node or not
            temp = temp.next

        #Once temp locate the last node, then pre should be pointing to the node before the last given that it lags behind temp by one node
        #We would want to set the tail to reference to point to the node that pre points to
        self.tail = pre 

        #disconnect and pop off the last node
        self.tail.next = None 

        #update the length of the node
        self.length -= 1

        #handling the head and tail references for the case when we pop off the only node in the LL
        if self.length == 0:
            self.head = None
            self.tail = None 

        return temp #We want to return the node that pops off

    def prepend(self, value):
        new_node=Node(value)

        #if LL is empty
        if self.length == 0:
            self.head = new_node
            self.tail = new_node
        else: #if LL is not empty
            new_node.next = self.head
            self.head = new_node

        self.length += 1

        return True #author plans to use this method for some boolean operation
    
    def pop_first(self):
        #Case where we have an empty LinkedList
        #I guess it makes sense for this to return None
        #In general when we are popping anything then we would like to reutn whatever it is that we pop
        #Int his csae if we do't have anything to pop, because the linkedlist itlsef is empty then it makes sense that we are returning None
        if self.length == 0:
            return None
        
        #In general we would want to use else: after an if statement
        #I think in this case, we could still use else statement. However, all the line of items that follow will need to be wrapped under the else statement.
        #Ultimately I thikn else statment in this scenarios is purely optional

        #I guess we found one reason why we may not want to use the else statement after all, because either we need to put all those lines under the else statement or even just one line, but it looks silly to have those lilnes be written with one line be indented inside an else statement
        # else: 
        
        temp = self.head
        self.head = self.head.next
        temp.next = None

        #Handling the tail. 
        #If performing pop first lead to the removal of the only node that our LinkedList has then at this point the self.head is taken care of, because it points to a None
        #However, the tail is still point to the node that we just poppped
        #Ultimately I don't think this is that big of a deal, but for correctness I think we should also free up the self.tail and make sure it points to none
        if self.length ==0:
            self.tail =None

        return temp #In general we want to reutn whatever it is that we pop. In this scenario we may be interested to know the value of the node, however in general we would like to return stuff that we pop in commpletion, that is the node. We could always call the value of the node if we want, but we ddon't want to reutn just node.value, we would like to reutnr the entire node. It's just best practice
    

```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class Stack:
    def __init__(self, value):
        new_node = Node(value)

        #Unlike the LinkedList where we have the head and tail reference
        #In stakcs we call them top and bottom, specifically the head would be top, and the tail would be the bottom
        #Recall that, stack uses the LIFO mechanism
        #LAst in, first out
        # self.head = new_node
        self.top = new_node

        #While it's ture that the tail reference would be equivalent to the bottom in stacks, what's also true is that Stacks don't have the need to track the bottom because it operates using the LIFO mechanism.
        #Including the bottom woul simply  a memory overhead that we don't need
        # self.tail = new_node
        # self.bottom = new_node

        #While we care about the length of the linkedlist, in the case of stakcs that is oriented vertically, we don't call the lenth as length.. Instead we refer to the length as height
        # self.length = 1
        self.height = 1

    def print_stack(self):
        #below to print for LinkedList
        # temp = self.head

        # while temp is not None:
        #     print(temp.value)
        #     temp = temp.next

        temp = self.top

        while temp is not None:
            print(temp.value)
            temp = temp.next

    def push(self, value):
        new_node = Node(value)

        if self.height == 0:
            self.top = new_node

            #Since we don't relly use the self.bottom for stakcs, given that stacks operate by LIFO, then introducing the self.bottom just add a memory overhead that we don't need
        else:
            new_node.next = self.top
            self.top = new_node
        self.height += 1

        return True #the author will use the method in some boolean operation
    
    def pop(self):
        if self.height ==0:
            return None
        
        temp = self.top
        self.top = self.top.next
        temp.next = None 
        self.height -= 1

        return temp
```

```python
my_stack = Stack(4)

my_stack.push(2)
my_stack.push(1)

my_stack.print_stack()
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class Queue:
    def __init__(self, value):
        new_node = Node(value)
        self.first = new_node
        self.last = new_node
        self.length = 1

    def print_queue(self):
        temp = self.first
        while temp is not None:
            print(temp.value)
            temp = temp.next

    def enqueue(self, value):
        new_node = Node(value)

        #case when the queue is empty
        if self.length == 0:
            self.first = new_node
            self.last = new_node
        else:
            self.last.next = new_node
            self.last = new_node
        self.length += 1

    def dequeue(self):
        #case when it's an empty list
        if self.length == 0:
            return None
        
        temp = self.first
        if self.length == 1:
            self.first = None
            self.last = None
        else:
            self.first = self.first.next
            temp.next = None
        self.length -= 1 
        return temp
        
```

```python
my_queue = Queue(1)

print('Queue before enqueue(2):')
my_queue.print_queue()
```

```python
my_queue.enqueue(2)

print('Queue after enqueue(2):')
my_queue.print_queue()
```

```python
class HashTable:
    def __init__(self, size=7):
        self.data_map = [None] * size

    def __hash(self,key):
        my_hash = 0
        for letter in key:
            my_hash = (my_hash + ord(letter) * 23) % len(self.data_map)
        return my_hash
    
    def print_table(self):
        for i, val in enumerate(self.data_map):
            print(i, ': ', val)
            
```

```python
my_hash_table = HashTable()
```

```python
my_hash_table.print_table()
```

```python
def item_in_common(list1, list2):
    for i in list1:
        for j in list2:
            if i == j:
                print('Common values:', list1[i])
                return True
    return False
```
