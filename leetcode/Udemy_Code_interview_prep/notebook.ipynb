```python
def print_items(n):
    for i in range(n):
        print(i)
```

```python
print_items(10)
```

```python
def print_items(n):
    for i in range(n):
        print(i)
    for j in range(n):
        print(j)
```

```python
print_items(10)
```

```python
def print_items(n):
    for i in range(n):
        for j in range(n):
            print(i,j)
```

```python
#O(n^2), nested loop
print_items(3)
```

```python
class Cookie:
    def __init__(self, color): #constructor
        self.color = color
    
    def get_color(self):
        return self.color
    
    def set_color(self, color):
        self.color=color

```

```python
cookie_one = Cookie('green')
```

```python
cookie_two = Cookie('blue')
```

```python
print('Cookie one is', cookie_one.get_color())
```

```python
print('Cookie two is', cookie_two.get_color())
```

```python
cookie_one.set_color('yellow')
```

```python
print('\nCookie one is now', cookie_one.get_color())
print('Cookie two is still', cookie_two.get_color())
```
