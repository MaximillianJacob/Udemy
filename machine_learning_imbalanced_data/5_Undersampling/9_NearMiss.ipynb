```python
df = pd.read_csv('../kdd2004.csv')
df.head()
```

```python
X = df.drop('target', axis=1)
X.columns
```

```python
y=df.target
y.value_counts(normalize=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3, random_state=0)
```

```python
nm1 = NearMiss(
    
    sampling_strategy='auto',
    version=1,
    n_neighbors=3,
    n_neighbors_ver3=3,
    n_jobs=-1,
)

nm2 = NearMiss(
    
    sampling_strategy='auto',
    version=2,
    n_neighbors=3,
    n_neighbors_ver3=3,
    n_jobs=-1,
)

nm3 = NearMiss(
    
    sampling_strategy='auto',
    version=3,
    n_neighbors=3,
    n_neighbors_ver3=3,
    n_jobs=-1,
)
```

```python
X_resampled, y_resampled = nm1.fit_resample(X_train, y_train)
X_resampled2, y_resampled2 = nm2.fit_resample(X_train, y_train)
X_resampled3, y_resampled3 = nm3.fit_resample(X_train, y_train)
```

```python
Scatterplot(X_train, y_train, X_resampled, y_resampled)
Scatterplot(X_resampled2, y_resampled2,X_resampled3,y_resampled3)
```
