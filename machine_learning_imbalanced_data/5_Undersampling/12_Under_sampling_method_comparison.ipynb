```python
from sklearn.preprocessing import MinMaxScaler
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=39,
    max_depth=3,
    n_jobs=-1
)
```

```python
from sklearn.metrics import roc_auc_score
```

```python
def RunClassifier(classifier, X_train, X_test, y_train, y_test):
    '''
    Ex:

    rf = RandomForestClassifier()

    >>>RunClassifier(rf, X_train, X_test, y_train, y_test)
    '''

    classifier.fit(X_train, y_train)

    classifier_name = type(classifier).__name__

    print('Train Set')
    train_prob = classifier.predict_proba(X_train)[:,1]
    print(classifier_name,"'s Train set roc-auc: {}".format(roc_auc_score(y_train, train_prob)))

    print('Test set')
    test_prob = classifier.predict_proba(X_test)[:,1]
    print(f"{classifier_name}'s Test set roc-auc score: {roc_auc_score(y_test, test_prob)}")
    
    return roc_auc_score(y_test, test_prob)
```

```python
results_dict = {}
shapes_dict = {}

for dataset in dataset_ls:

    results_dict[dataset] ={}
    shapes_dict[dataset] = {}

    print(dataset)

    #Loading the dataset
    data = fetch_datasets()[dataset]
    
    X_train, X_test, y_train, y_test = train_test_split(
        data.data,
        data.target,
        test_size=0.3,
        random_state=0
    )

    scaler = MinMaxScaler().fit(X_train)
    X_train = scaler.transform(X_train)
    X_test = scaler.transform(X_test)

    roc = RunClassifier(rf, X_train, X_test, y_train, y_test)

    #store results

    results_dict[dataset]['full_data'] = roc
    shapes_dict['full_data'] = len(X_train)

    print()

    for undersampler in undersampler_dict.keys():
        print(undersampler)

        X_resampled, y_resampled = undersampler_dict[undersampler].fit_resample(X_train, y_train)
        
        roc = RunClassifier(rf, X_resampled, X_test, y_resampled, y_test)

        results_dict[dataset][undersampler] = roc
        shapes_dict[dataset][undersampler] = len(X_resampled)
        print()

    print()

```
