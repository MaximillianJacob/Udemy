```python
def twoSum(nums, target):
    num_index_dict = {}

    for index, value in enumerate(nums):
        # print(num_index_dict)
        complement = target - value

        if complement in num_index_dict:
            return index, num_index_dict[complement]
        
        num_index_dict[value] = index

    return None

```

```python
nums = [2,11,15,7]
target = 9
twoSum(nums, target)
```
