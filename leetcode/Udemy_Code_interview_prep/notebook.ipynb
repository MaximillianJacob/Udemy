```python
class Cookie:
    def __init__(self, color): #constructor
        self.color = color
    
    def get_color(self):
        return self.color
    
    def set_color(self, color):
        self.color=color

```

```python
cookie_one = Cookie('green')
```

```python
cookie_two = Cookie('blue')
```

```python
print('Cookie one is', cookie_one.get_color())
```

```python
print('Cookie two is', cookie_two.get_color())
```

```python
cookie_one.set_color('yellow')
```

```python
print('\nCookie one is now', cookie_one.get_color())
print('Cookie two is still', cookie_two.get_color())
```

```python
class Node:
    def __init__(self, value):
        #We know that a node has 2 fields, value and next
        self.value = value
        self.next = None
```

```python
class LinkedList:
    def __init__(self, value):
        # create new node and pass value to this node
        new_node = Node(value)

        # assign new node to head
        self.head = new_node

        # assign new node to tail
        self.tail = new_node

        #count the length of the node
        self.length = 1
```

```python
my_linked_list = LinkedList(4)
```

```python
dir(my_linked_list)
```

```python
dir(my_linked_list.head)
```

```python
print(my_linked_list.head.value)
```

```python
#LL Constructor

class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedList:
    def __init__(self, value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.length = 1
```

```python
#create an instance of a LinkedList
my_linked_list = LinkedList(4)
```

```python
print('Head:', my_linked_list.head.value)
print('Tail:', my_linked_list.tail.value)
print('Length:', my_linked_list.length)
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None



class LinkedList:
    def __init__(self, value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.length = 1

    def print_list(self):
        #this is a temporary folder, hence se don't have the self.temp
        temp = self.head

        while temp is not None:
            print(temp.value)
            temp = temp.next
```
