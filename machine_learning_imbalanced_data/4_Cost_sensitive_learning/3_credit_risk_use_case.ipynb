```python
from sklearn.linear_model import LogisticRegression
```

```python
logit = Pipeline([
    ('imputer_cat', imputer_categorical),
    ('rare_categories', rare_encoder),
    ('missing_indicators', missing_indicators),
    ('imputer_num', imputer_mean),
    ('encoder', onehot),
    ('model', LogisticRegression(
        penalty='l2',
        C=0.1,
        random_state=10
    )),
]).set_output(transform='pandas')
```

```python
dir(Pipeline)
```

```python
help(Pipeline.set_output)
```

```python
logit.fit(X_train, y_train)
```

# With Cost-sensitive Learning

```python
#the cost-sensitive learning component is introduced through the class_weight parameter
logit_csl = Pipeline([
    ('imputer_cat', imputer_categorical),
    ('rare_categories', rare_encoder),
    ('missing_indicators', missing_indicators),
    ('imputer_num', imputer_mean),
    ('encoder', onehot),
    ('model', LogisticRegression(
        penalty='l2',
        C=0.1,
        class_weight = {0:1, 1:10}, #introducing the cost-sensitive learning. Misclassifying class 0 will incur a cost of 1, Misclassifying class 1 will incur a cost of 10
        random_state=10
    ))
]).set_output(transform='pandas')
```

```python
logit_csl.fit(X_train, y_train)
```
