```python
param_grid = {
    'n_estimators' : [10,50,100],
    'max_depth': [None, 2, 3],
    'class_weight': [None, {-1:1, 1:10}, {-1:1, 1:100}] #cost optimization
}
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data.drop('target', axis=1),
    data.target,
    test_size=0.3,
    random_state=0
)
```

```python
search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    scoring='roc_auc',
    n_jobs=-1,
    cv=2,
)

search.fit(X_train, y_train)
```
