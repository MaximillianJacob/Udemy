```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self, value):
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.length = 1

    def append(self, value):
        new_node = Node(value)

        if self.length == 0:
            self.head = new_node
            self.tail = new_node
        
        self.tail.next = new_node
        self.tail = new_node

        #update the length
        self.length += 1

        return True

    def print_list(self):
        temp = self.head

        while temp != None:
            print(temp.value)
            temp = temp.next

    def pop(self):
        #handle case where empty list
        if self.length == 0:
            return None

        #handle case where LL contain more than one node
        temp = self.head
        pre = self.head

        while temp.next != None:
            pre = temp
            temp = temp.next   

        self.tail = pre
        self.tail.next = None 

        #updating the length
        self.length -= 1

        if self.length == 0:
            self.tail = None
            self.head = None

        return temp
    
    def prepend(self, value):
        new_node = Node(value)

        if self.length == 0:
            self.head = new_node
            self.tail = new_node

        else:
            new_node.next = self.head
            self.head = new_node

        self.length += 1

        return True
        
    def pop_first(self):
        if self.length == 0:
            return None
        
        temp = self.head
        self.head = self.head.next
        temp.next = None

        #updating the length
        self.length -= 1

        if self.length == 0:
            self.tail = None
        
        return temp

    
```

```python
my_linked_list = LinkedList(2)
my_linked_list.append(1)


# (2) Items - Returns 2 Node
print(my_linked_list.pop_first().value)
# (1) Item -  Returns 1 Node
print(my_linked_list.pop_first().value)
# (0) Items - Returns None
print(my_linked_list.pop_first())
```
