```python
def print_items(n):
    for i in range(n):
        print(i)
```

```python
print_items(10)
```

```python
def print_items(n):
    for i in range(n):
        print(i)
    for j in range(n):
        print(j)
```

```python
print_items(10)
```

```python
def print_items(n):
    for i in range(n):
        for j in range(n):
            print(i,j)
```
