# Under-sample observations from the majority class

```python
y.value_counts()
```

```python
#Median, 50%, shows that majority class have probability values ~ 0.37. 
#At least half of the observations from the majority class have probability <0.37
probs.describe()
```

```python
#Here using 0.3 as an arbitrary threshold value

condition_0 =  (y==0) & (probs['p(0)'] <= 0.3)
condition_0
```

```python
condition_0.sum()
```

```python
probs.columns
```

```python
#Make sure to wrap the condition in brackets
condition_1 = (y==1) & (probs['p(1)'] <= 0.3)
```

```python
filtered_0 = X[condition_0]
filtered_0.head()
filtered_0.shape
```

```python
filtered_1 = X[condition_1]
filtered_1.shape
```

```python
X.columns
```

```python
test = pd.concat([X, y, (y + 3*condition_0)], axis=1)
test.columns = ['varA', 'varB', 'y', 'y2']
test.head()
```

```python
#In here, adding  the 3* condition_0, is perofrmed simply so that we could distinguish the data point that satisfies our condition_0
#The 3 is arbitrary, however it needs to be at least times 3, anything less the unique value remains only 0,1,2 and the data point would skew with class distribution for other class
test.y.unique(), test.y2.unique()
```

```python
class_ = 0 

#This is the number of samples we are interested in, from the minority class
n_samples = balance[2]

majority_samples = balance[class_]

#We are trying to see how many percent of the majority class data would amount to the number of the target minority class
perc = n_samples/majority_samples


#We know the percentage of the number of the majority class 0 that would amount to the minorty class
#However what will happen is that, we will be obtaining the percentage of the majority class, specifically the percentage of the upper percentile from the majority class with the highest probability

perc = (1 - perc) * 100
print(perc)
```
