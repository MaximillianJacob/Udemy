```python
def RunClassifier(classifier, X_train, X_test, y_train, y_test):
    '''
    Ex:

    rf = RandomForestClassifier()

    >>>RunClassifier(rf, X_train, X_test, y_train, y_test)
    '''

    classifier.fit(X_train, y_train)

    classifier_name = type(classifier).__name__

    print('Train Set')
    train_prob = classifier.predict_proba(X_train)[:,1]
    print(classifier_name,"'s Train set roc-auc: {}".format(roc_auc_score(y_train, train_prob)))

    print('Test set')
    test_prob = classifier.predict_proba(X_test)[:,1]
    print(f"{classifier_name}'s Test set roc-auc score: {roc_auc_score(y_test, test_prob)}")
```

```python
from sklearn.metrics import roc_auc_score
```

```python
RunClassifier(rf, X_train, X_test, y_train, y_test)
```

```python
def inspect_classifier_attributes(classifier):
    """
    Prints all attributes and their values for the given classifier.
    """

    for attr in dir(classifier):  # Loop through all attributes
        try:
            value = getattr(classifier, attr)  # Get attribute value
            print(f"{attr}: {value}\n")
        except Exception as e:
            print(f"{attr}: Could not retrieve value ({e})\n")
```
